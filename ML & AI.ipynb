```python
#This is used to clean the columns description of extra spaces as seen in the initial call compared to the new call after the cleaning.
#It works only for strings...
dataset.columns = dataset.columns.str.strip()
```

```python
dataset['education'] = dataset['education'].str.strip()
```

```python
dataset['self_employed'] = dataset['self_employed'].str.strip()
dataset['loan_status'] = dataset['loan_status'].str.strip()
```

```python
dataset
```

```python
# It is always good to convert our string values to numeric for computation purposes.

dataset['education'] = dataset['education'].map({'Graduate': 1, 'Not Graduate': 0})
dataset['self_employed'] = dataset['self_employed'].map({'Yes': 1, 'No': 0})
dataset['loan_status'] = dataset['loan_status'].map({'Approved': 1, 'Rejected': 0})
```

```python
dataset
```

```python
# Creating a total asset by summing together all of the assets in the data.

dataset['total_assets'] = (dataset['residential_assets_value'] + 
                           dataset['commercial_assets_value'] + 
                           dataset['luxury_assets_value'] + 
                           dataset['bank_asset_value'])
```

```python
dataset
```

```python
# calculating the financial capability of the users after the deduction of the loan amount.
dataset['financial_capability'] = (
    dataset['income_annum'] +
    dataset['total_assets'] -
    dataset['loan_amount']
)
```

# EDA, Visualization & Collaboration

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\USER\Downloads\SalesData.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df = df.drop_duplicates()
```

```python
df
```

```python
# Sales by Region
df.groupby('Region')['Sales'].sum().plot(kind='bar')
plt.title("Sales by Region")
plt.show()
```

```python
# Sales by Category
df.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.title("Sales by Category")
plt.show()
```

```python
# Top Products
df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title("Top 10 Products")
plt.show()
```

```python
# Sales by Segment
df.groupby('Segment')['Sales'].sum().plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales by Customer Segment")
plt.ylabel('')
plt.show()
```

```python
#Top 10 States by Sales
df.groupby('State')['Sales'].sum().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title("Top 10 States by Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
#Top 10 Customers by Sales
df.groupby('Customer Name')['Sales'].sum().sort_values(ascending=False).head(10).plot(kind='bar')
plt.title("Top 10 Customers by Sales")
plt.xticks(rotation=75)
plt.show()
```

```python
# Sales by Ship Mode
df.groupby('Ship Mode')['Sales'].sum().plot(kind='bar')
plt.title("Sales by Ship Mode")
plt.show()
```
